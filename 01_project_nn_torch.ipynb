```python
# load libraries
import torch
import torch.nn as nn

# complete real neural network
class PytorchNeuralNetwork(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(input_dim, 128),

            nn.ReLU(),

            nn.Linear(128, 64),

            nn.ReLU(),

            nn.Linear(64, 32),

            nn.ReLU(),

            nn.Linear(32, 1)
        )

    def forward(self, x):

        return self.network(x)

# no.29
```

## Binary Classification Project

#### first PyTorch neural network.

```python
# load sklearn dataset
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=7,
    n_redundant=2,
    n_classes=2,
    random_state=42
)

# convert to tensors
X = torch.tensor(
    X, 
    dtype=torch.float32
)

y = torch.tensor(
    y, 
    dtype=torch.float32
).reshape(-1, 1)
```

```python
# check shape & size of the dataset
print(X.shape)
print(y.shape)
```

```python
# create pytorch neural network
model = PytorchNeuralNetwork(
    input_dim=X.shape[1] # shape
)

print(model)
```

```python
# define loss & optimizer
criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
# train the model (training loop)

# number of epochs
epochs = 5

for epoch in range(epochs):

    model.train()

    optimizer.zero_grad()

    logits = model(X)

    loss = criterion(
        logits, y
    )

    loss.backward()

    optimizer.step()

    if (epoch + 1 ) % 10 == 0:

        print(
            f"Epoch [{epoch + 1}]/{epochs} "
            f"Loss: {loss.item():.4f}"
        )
```

```python
# make predictions

# evaluate model
# from torch import sigmoid
model.eval()

with torch.no_grad():

    logits = model(X)

    probabilities = torch.sigmoid(
        logits
    )

    predictions = (
        probabilities >= 0.5
    ).float()

# print probabilities
print(probabilities[:10])

# print predictions
print(predictions[:10])
```

```python
# calculate model accuracy
accuracy = (
    predictions == y
).float().mean()

print(
    f"Accuracy: {accuracy.item():.4f}"
)

# next lesson 4
```
