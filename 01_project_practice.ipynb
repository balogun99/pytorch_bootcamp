```python
import torch
import torch.nn as nn

class ExerciseModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(5, 16)
        self.layer2 = nn.Linear(16, 8)
        self.output = nn.Linear(8, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = torch.sigmoid(x)

        x = self.layer2(x)
        x = torch.sigmoid(x)

        x = self.output(x)

        return x
```

```python
# import libraries
import torch
import torch.nn as nn

class NeuralNetwork(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(input_dim, 64),

            nn.ReLU(),

            nn.Linear(64, 32),

            nn.ReLU(),

            nn.Linear(32, 1)
        )

    def forward(self, x):

        return self.network(x)
```

```python

from sklearn.datasets import make_classification

X, y = make_classification(
    n_features=10,
    random_state=42
)

X = torch.tensor(
    X,
    dtype=torch.float32
)

y = torch.tensor(
    y,
    dtype=torch.float32
).reshape(-1, 1)
```

```python
print(X.shape)
print(y.shape)
```

```python
model = NeuralNetwork(
    input_dim=X.shape[1]
)

print(model)
```

```python
criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
# train the model
epochs = 100

for epoch in range(epochs):

    model.train()

    optimizer.zero_grad()

    logits = model(X)

    loss = criterion(
        logits, y
    )

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 10 == 0:

        print(
            f"Epoch [{epoch + 1}]/{epochs} "
            f"Loss: {loss.item():.4f}"
        )
```

```python
# model evaluation
model.eval()

with torch.no_grad():

    logits = model(X)

    probabilities = torch.sigmoid(
        logits
    )

    predictions = (
        probabilities >= 0.5
    ).float()

print(probabilities[:10])
print(predictions[:10])
```

```python
# model accuracy
accuracy = (
    predictions == y
).float().mean()

print(
    f"Accuracy: {accuracy.item():.4f}"
)
```
