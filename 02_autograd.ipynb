```python
# autograd - PyTorch's automatic differentiation engine.
# This is what allows PyTorch to calculate the gradients needed to train neural networks.

# using requires_grad=True
import torch

x = torch.tensor(
    3.0,
    requires_grad=True
)

print(x)
```

```python
# calculate the function
# y=x 
# 2
 
# The derivative is:
# dx
# dy=2x
y = x ** 2

print(y)
```

```python
# calculate gradient
y.backward()
```

```python
print(x.grad)
```

```python
# y=3x2+2x+1
import torch

x = torch.tensor(
    2.0,
    requires_grad=True
)

y = 3 * x ** 2 + 2 * x + 1

y.backward()

print("y =", y)
```

```python
# computational graph
# When you perform operations on tensors with gradient tracking enabled, PyTorch builds a computational graph.
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2
z = 2 * y

z.backward()
print(z)
```

```python
# complex examples
x = torch.tensor(
    2.0,
    requires_grad=True
)

a = x ** 2
b = 3 * a
c = b + 5

c.backward()
print(x.grad)
```

```python
# multiple variables
# y=wx+b
x = torch.tensor(2.0)

w = torch.tensor(
    3.0,
    requires_grad=True
)

b = torch.tensor(
    1.0,
    requires_grad=True
)

y = w * x + b

print(y)
```

```python
y.backward()
```

```python
print("dw:", w.grad)
print("db:", b.grad)
```

# # gradient descent
# # Suppose we have a parameter: w
# # and its gradient is:
# # gradient = 4
# # A simple gradient descent update is:
# w 
# new
# ​	
#  =w−η 
# ∂w
# ∂L
# ​	
 
# where:
# w = current parameter
# L = loss
# η = learning rate
# gradient = derivative of loss with respect to parameter
# For example:
# w = 10
# learning rate = 0.1
# gradient = 4
# Then:
# w 
# new
# ​	
#  =10−(0.1)(4)
# w 
# new
# ​	
#  =9.6
# So the parameter moved in the direction that should reduce the loss.

```python
# manually performing Gradient Descent
x = torch.tensor(
    2.0,
    requires_grad=True
)

w = torch.tensor(
    5.0,
    requires_grad=True
)

y = w * w
```

```python
target = torch.tensor(20.0)

loss = (y - target) ** 2

print("Prediction: ", y)
print("Loss: ", loss)
```

```python
loss.backward()
```

```python
print("Gradient: ", w.grad)
```

```python
# update w
with torch.no_grad(): 
    w -= 0.01 * w.grad

print(w)
```

```python
# .detach()
# detach() creates a tensor that is disconnected from the computational graph.
# This is useful when you want to work with a tensor's value without continuing gradient tracking through it.
x = torch.tensor(
    3.0,
    requires_grad=True
)

y = x ** 2

z = y.detach()
```

```python
print(y.requires_grad)
print(z.requires_grad)
```

| Concept              | Purpose                         |
| -------------------- | ------------------------------- |
| Tensor               | Stores numerical data           |
| `requires_grad=True` | Track operations for gradients  |
| Computational graph  | Records mathematical operations |
| `.backward()`        | Calculates gradients            |
| `.grad`              | Stores calculated gradients     |
| `torch.no_grad()`    | Disable gradient tracking       |
| `.detach()`          | Disconnect tensor from graph    |
| `zero_grad()`        | Clear old gradients             |
| `optimizer.step()`   | Update parameters               |
