```python
# import libraries
# Pytorch Neural Network Architecture
import torch
import torch.nn as nn

class Exercise03NN(nn.Module):

    def __init__(self, input_dim):
        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(input_dim, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.30),

            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.20),

            nn.Linear(64, 32),
            nn.ReLU(),


            nn.Linear(32, 1)
        )

    def forward(self, x):

        return self.network(x)
```

```python
# load sklearn dataset

from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=7,
    n_redundant=2,
    n_classes=2,
    random_state=42
)

# convert to tensors
X = torch.tensor(
    X,
    dtype=torch.float32
)

y = torch.tensor(
    y,
    dtype=torch.float32
).reshape(-1, 1)
```

```python
# shape & size of the dataset
print(X.shape)
print(y.shape)
```

```python
# create the model
model = Exercise03NN(
    input_dim=X.shape[1]
)

print(model)
```

```python
# loss & optimizer
criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
# model training
epochs = 100

for epoch in range(epochs):

    model.train()

    optimizer.zero_grad()

    logits = model(X)

    loss = criterion(
        logits, y
    )

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 10 == 0:

        print(
            f"Epoch [{epoch + 1}]/{epochs} "
            f"Loss: {loss.item():.4f}"
        )
```

```python
# model evaluation

model.eval()

with torch.no_grad():

    logits = model(X)

    probabilities = torch.sigmoid(
        logits
    )

    predictions = (
        probabilities >= 0.5
    )

    # print probabilities
    print(probabilities[:10])

    # print predictions
    print(predictions[:10])
```

```python
# model accuracy
accuracy = (
    predictions == y
).float().mean()

print(
    f"Accuracy: {accuracy.item():.4f}"
)
```

```python
# save the model
torch.save(
    model.state_dict(),
    "model.pth"
)
```

```python
# load the model
model.load_state_dict(
    torch.load(
        "model.pth",
        weights_only=True
    )
)
```
