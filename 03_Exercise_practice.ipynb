```python
# import libraries
import torch
import torch.nn as nn

class Ex03Practice(nn.Module):

    def __init__(self, input_dim):
        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(input_dim, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.30),

            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.20),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1)

        )

    def forward(self, x):

        return self.network(x)
```

```python
# load sklearn dataset
from sklearn.datasets import make_regression

X, y = make_regression(
    n_samples=100,
    n_features=20,
    n_informative=5,
    n_targets=1,
    random_state=42
)

# convert to tensors
X = torch.tensor(
    X,
    dtype=torch.float32
)

y = torch.tensor(
    y,
    dtype=torch.float32
).reshape(-1, 1)
```

```python
# size & shape of the dataset
print(X.shape)
print(y.shape)
```

```python
# create the model
model = Ex03Practice(
    input_dim=X.shape[1]
)

print(model)
```

```python
# loss & optimizer
criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.05
)
```

```python
# model training
epochs = 100

for epoch in range(epochs):

    model.train()

    optimizer.zero_grad()

    logits = model(X)

    loss = criterion(
        logits, y
    )

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 10 == 0:

        print(
            f"Epoch: [{epoch + 1}]/{epoch} "
            f"Loss: {loss.item():.4f}"
        )
```

```python
# model evaluation

model.eval()

with torch.no_grad():

    logits = model(X)

    probabilities = torch.sigmoid(
        logits
    )

    predictions = (
        probabilities >= 0.5
    )

    # print probabilities
    print(probabilities[:5])

    # print predictions
    print(predictions[:5])
```

```python
# model accuracy
accuracy = (
    predictions == y
).float().mean()

print(
    f"Accuracy: {accuracy.item():.4f}"
)
```

```python
# save the model
torch.save(
    model.state_dict(),
    "03_EX_Practice.pth"
)
```

```python
# load the model
model.load_state_dict(
    torch.load(
        "03_EX_Practice.pth",
        weights_only=True
    )
)
```
