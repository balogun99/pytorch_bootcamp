## NN Module
## nn.Module is the base class for neural network models in Pytorch

```python
import torch
import torch.nn as nn

class MyModel(nn.Module):

    def __init__(self):
        super().__init__()

    def forward(self, x):
        return self.layer(x)

# __init_() -> is used to define the components of the network
# forward() -> this method is used to defines how data flows through the network
```

```python
# simple linear model
import torch
import torch.nn as nn

class LinearModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.linear = nn.Linear( # Pytorch Layers
            in_features=1,
            out_features=1
        )

    def forward(self, x):
        return self.linear(x)

model = LinearModel()
print(model)
```

```python
# model parameters
model = LinearModel()

for parameter in model.parameters():
    print(parameter)

print(model.linear.weight)
print(model.linear.bias)
```

```python
# number of parameters
total_parameters = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(total_parameters) # -> nn.Linear(1, 1)
```

```python
# forward pass
X = torch.tensor(
    [[1.0],
    [2.0],
    [3.0],
    [4.0]]
)

predictions = model(X)
print(predictions)
```

-> Rectified Linear Unit
-> Mathematically:
-> ReLU(x)=max(0,x)

```python
# example
x = torch.tensor(
    [-3., -1., 0., 2., 5]
)

print(relu(x))
```

## Sigmoid
Sigmoid converts values approximately into:
-> 0, 1

Its output is between:
0<sigmoid(x)<1

```python
# In Pytorch
sigmoid = nn.Sigmoid()
```

```python
# example
from torch import sigmoid


x = torch.tensor(
    [-2., 0., 2.]
)

print(sigmoid(x))
```

```python
# multi-layer neural network
class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64)
        self.layer2 = nn.Linear(64, 32)
        self.output = nn.Linear(32, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = torch.relu(x)

        x = self.layer2(x)
        x = torch.relu(x)

        x = self.output(x)

        return x

```

```python
model = NeuralNetwork()
print(model)
```

```python
# passing data through the Network
X = torch.rand(32, 10) # 32 Samples, 10 Features
```

```python
output = model(X)
print(output.shape)
```

```python
# binary classification
# Let's make the model suitable for binary classification.
class BinaryClassifier(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64)
        self.layer2 = nn.Linear(64, 32)
        self.output = nn.Linear(32, 1)

    def forward(self, x):

        x = torch.relu(self.layer1(x))
        x = torch.relu(self.layer2(x))

        x = torch.sigmoid(
            self.output(x)
        )

        return x

# model
model = BinaryClassifier()

# pass data through the network
X = torch.rand(32, 10)

# print predictions
predictions = model(X)
```

```python
print(predictions.shape)
print(predictions[:5])
```

```python
# prediction threshold
# For binary classification, we can convert probabilities to classes.
predictions = model(X)

classes = (predictions >= 0.5).float()

print(classes[:5])
```
