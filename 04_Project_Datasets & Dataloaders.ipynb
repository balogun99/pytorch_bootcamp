```python
# import libraries
import torch
import torch.nn as nn

# generate dataset

from sklearn.datasets import make_classification
# from torch.utils.data import TensorDataset
# from torch.utils.data import random_split
# from torch.utils.data import DataLoader
from torch.utils.data import (
    Dataset,
    TensorDataset,
    DataLoader,
    random_split
)

X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=7,
    n_redundant=2,
    n_classes=2,
    random_state=42
)
```

```python
# convert to tensors
X = torch.tensor(
    X, 
    dtype=torch.float32
)

y = torch.tensor(
    y, 
    dtype=torch.float32
).reshape(-1, 1)
```

```python
# create dataset
dataset = TensorDataset(
    X, 
    y
)
```

```python
# split dataset (70% train, 15% test, 15% validation)
train_size = int(0.70 * len(dataset))
val_size = int(0.15 * len(dataset))
test_size = (
    len(dataset)
    - train_size
    - val_size
)
```

```python
# use reproducible split
generator = torch.Generator().manual_seed(42)

train_dataset, val_dataset, test_dataset = random_split(
    dataset,
    [train_size, val_size, test_size],
    generator=generator
)
```

```python
# create DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=32,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0
)
```

```python
# verify the pipeline
# before training, always inspect the data

X_batch, y_batch = next(
    iter(train_loader)
)

print("X_Batch Shape: ", X_batch.shape)
print("Y_Batch Shape: ", y_batch.shape)
```
