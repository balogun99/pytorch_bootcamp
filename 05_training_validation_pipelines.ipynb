```python
# Understanding Underfitting

# Imagine:

# Training accuracy = 65%
# Validation accuracy = 64%

# Both are poor.

# This could indicate underfitting.

# The model isn't learning the underlying patterns sufficiently.

# Possible solutions:

# More training
# Larger model
# Better features
# Better preprocessing
# Reduce excessive regularization
```

```python
# Understanding Overfitting

# Imagine:

# Epoch 10

# Training Accuracy = 95%
# Validation Accuracy = 75%

# The model performs extremely well on training data but poorly on unseen data.

# That's a classic sign of:

# Overfitting

# The model has learned the training data too specifically.

# 23. Typical Overfitting Pattern

# You may see:

# Training Loss
#      ↓
# 0.60
# 0.45
# 0.30
# 0.20
# 0.10
# 0.05

# while:

# Validation Loss
#      ↓
# 0.62
# 0.50
# 0.44
# 0.43
# 0.47
# 0.58
# 0.72

# The validation loss initially improves, then starts getting worse.

# That's a major warning sign.
```

```python
# 25. Early Stopping

# One solution is Early Stopping.

# The idea:

# Stop training when validation performance stops improving.

# For example:

# Epoch 1 → Val Loss 0.60
# Epoch 2 → Val Loss 0.52
# Epoch 3 → Val Loss 0.47
# Epoch 4 → Val Loss 0.44
# Epoch 5 → Val Loss 0.43
# Epoch 6 → Val Loss 0.44
# Epoch 7 → Val Loss 0.46

# The best model was around:

# Epoch 5

# We don't necessarily want to keep training indefinitely.

# 25. Patience

# A common early stopping parameter is:

# patience = 5

# Meaning:

# Allow up to 5 epochs without improvement before stopping.

# For example:

# Best validation loss
#        ↓
# Epoch 20
#        ↓
# No improvement
# Epoch 21
# Epoch 22
# Epoch 23
# Epoch 24
# Epoch 25
#        ↓
# STOP
```
