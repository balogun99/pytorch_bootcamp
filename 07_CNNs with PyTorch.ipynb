CNN stands for:

Convolutional Neural Network

It is a neural network architecture particularly suited to data with spatial structure, especially images.

Image Tensor Shape in PyTorch

This is extremely important.

PyTorch normally uses:

[Batch, Channels, Height, Width]

What Is Convolution?

The basic idea is that a small learnable filter moves across an image.

What Does a Filter Learn?

Early CNN layers may learn features resembling:

Vertical edges
Horizontal edges
Diagonal edges
Corners
Textures

Later layers can combine these into more complex patterns.

Conceptually:

Layer 1
 ↓
Edges

Layer 2
 ↓
Textures / simple shapes

Layer 3
 ↓
Object parts

Layer 4
 ↓
Higher-level representations

In PyTorch:

nn.Conv2d()

is the main convolutional layer for 2D images.

```python
# Basic example:

# nn.Conv2d(
#     in_channels=3,
#     out_channels=32,
#     kernel_size=3
# )

# This means:

# Input channels  = 3
# Output channels = 32
# Kernel size     = 3 × 3
```

What Does out_channels=32 Mean?

It means the convolution layer learns:

32 different filters

Each filter can learn a different feature.

Therefore:

Input
[3, 32, 32]

        ↓

Conv2d(
    3 → 32
)

        ↓

Output
[32, H, W]

The spatial dimensions depend on kernel size, stride, and padding.

* Kernel Size

Common choices include:

3 × 3
5 × 5

For example:

nn.Conv2d(
    3,
    32,
    kernel_size=3
)

means:

3 × 3 convolution kernel

A 3×3 kernel is extremely common in CNN architectures.

* Stride

Stride controls how far the kernel moves each time.

For example:

stride=1

means:

Move one pixel at a time.

Whereas:

stride=2

means:

Move two pixels at a time.

Increasing stride generally reduces spatial dimensions.

* Padding

Padding adds pixels around the boundary.

For example:

padding=1

with:

kernel_size=3

is commonly used to preserve height and width when:

stride = 1

For example:

nn.Conv2d(
    3,
    32,
    kernel_size=3,
    stride=1,
    padding=1
)

* Max Pooling

Another important CNN operation is:

nn.MaxPool2d()

For example:

nn.MaxPool2d(
    kernel_size=2
)

A 2×2 max-pooling layer typically reduces:

32 × 32

to:

16 × 16

when the stride defaults to 2.

* CNN Architecture

A basic CNN might look like:

Input Image
     ↓
Conv2D
     ↓
ReLU
     ↓
MaxPool
     ↓
Conv2D
     ↓
ReLU
     ↓
MaxPool
     ↓
Flatten
     ↓
Linear
     ↓
Output
